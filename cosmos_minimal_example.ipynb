## Simple Usage - Just 3 Lines!

````python
import cosmos

# Single line to create predictor - replaces complex data loading
model = cosmos.predictor('desert-hws')

# Single line to get odor concentration
concentration = model.step_update(1.0, 0.5)
print(f"Odor concentration: {concentration}")
````

## Available Models

````python
# List all available pre-trained models
models = cosmos.list_available_models()
for model in models:
    print(f"  {model['name']}: {model['description']}")
````

## Real-time Trajectory Simulation

````python
import numpy as np
import matplotlib.pyplot as plt

# Create model
model = cosmos.predictor('desert-hws')

# Define trajectory
dt = 0.005
tsim = np.arange(0, 10, dt)
x_pos = np.sin(tsim * 2 * np.pi * 0.5) + 2
y_pos = np.sin(tsim * 2 * np.pi * 4)

# Simulate odor experience along trajectory
odors = []
for x, y in zip(x_pos, y_pos):
    concentration = model.step_update(x, y, dt)
    odors.append(concentration)

print(f"Simulated {len(odors)} time steps")
````

## Visualization

````python
# Plot results
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

# Trajectory
axes[0, 0].plot(x_pos, y_pos)
axes[0, 0].set_title('Agent Trajectory')
axes[0, 0].set_xlabel('X Position')
axes[0, 0].set_ylabel('Y Position')
axes[0, 0].axis('equal')

# X position over time
axes[0, 1].plot(tsim, x_pos)
axes[0, 1].set_title('X Position vs Time')
axes[0, 1].set_xlabel('Time (s)')
axes[0, 1].set_ylabel('X Position')

# Y position over time
axes[1, 0].plot(tsim, y_pos)
axes[1, 0].set_title('Y Position vs Time')
axes[1, 0].set_xlabel('Time (s)')
axes[1, 0].set_ylabel('Y Position')

# Odor concentration over time
axes[1, 1].plot(tsim, odors)
axes[1, 1].set_title('Odor Concentration Experience')
axes[1, 1].set_xlabel('Time (s)')
axes[1, 1].set_ylabel('Concentration')

plt.tight_layout()
plt.show()
````

## Statistics

````python
print(f"Simulation statistics:")
print(f"  Total time steps: {len(odors)}")
print(f"  Mean concentration: {np.mean(odors):.4f}")
print(f"  Max concentration: {np.max(odors):.4f}")
print(f"  Min concentration: {np.min(odors):.4f}")
print(f"  Std concentration: {np.std(odors):.4f}")

# Count whiff events (above background)
background_level = 0.6
whiff_mask = np.array(odors) > background_level + 0.1
whiff_count = np.sum(whiff_mask)
whiff_percentage = (whiff_count / len(odors)) * 100

print(f"  Whiff events: {whiff_count} ({whiff_percentage:.1f}% of time)")
````
